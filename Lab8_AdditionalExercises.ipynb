```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_available = True

class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title, author):
        self.books.append(Book(title, author))

    def borrow_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if book.is_available:
                    book.is_available = False
                    return f"Successfully borrowed '{book.title}'."
                return f"'{book.title}' is already borrowed."
        return "Book not found."

    def return_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                book.is_available = True
                return f"'{book.title}' has been returned."
        return "Book not found."

# Execution
lib = Library()
lib.add_book("Python Programming", "A. Sweigart")
print(lib.borrow_book("Python Programming"))
print(lib.borrow_book("Python Programming")) # Test double borrowing
print(lib.return_book("Python Programming"))
```

```python
class Employee:
    def __init__(self, name, rate, hours):
        if rate < 0 or hours < 0:
            raise ValueError("Rate and hours must be non-negative.")
        self.name = name
        self.rate = rate
        self.hours = hours

    def calculate_pay(self):
        overtime_hours = max(0, self.hours - 160)
        base_hours = self.hours - overtime_hours
        base_pay = base_hours * self.rate
        overtime_pay = overtime_hours * (self.rate * 1.5)
        return base_pay, overtime_pay

# Execution
try:
    emp = Employee("Michael", 500, 170)
    base, ot = emp.calculate_pay()
    print(f"Employee: {emp.name}")
    print(f"Base Salary: P{base}")
    print(f"Overtime Pay: P{ot}")
    print(f"Total Monthly Salary: P{base + ot}")
except ValueError as e:
    print(f"Error: {e}")
```

```python
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

class ShoppingCart:
    def __init__(self):
        self.cart = []

    def add_to_cart(self, product, quantity):
        if quantity <= product.stock:
            self.cart.append((product, quantity))
            product.stock -= quantity
            print(f"Added {quantity} x {product.name}")
        else:
            print(f"Insufficient stock for {product.name}!")

    def get_total(self):
        return sum(item[0].price * item[1] for item in self.cart)

# Execution
laptop = Product("Laptop", 45000, 5)
cart = ShoppingCart()
cart.add_to_cart(laptop, 2)
print(f"Current Cart Total: P{cart.get_total()}")
cart.add_to_cart(laptop, 10) # Test stock rejection
```
