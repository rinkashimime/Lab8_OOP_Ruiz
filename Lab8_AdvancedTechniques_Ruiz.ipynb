```python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int
    course: str

class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius  # Private attribute

    @property
    def fahrenheit(self):
        # Controlled access: converts C to F on the fly
        return (self._celsius * 9/5) + 32

# Test Output
s = Student("John Cruz", 20, "ECE")
t = Temperature(25)
print(f"Student Data: {s}")
print(f"25°C in Fahrenheit is: {t.fahrenheit}°F")
```
